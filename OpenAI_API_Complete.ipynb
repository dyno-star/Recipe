#<font color='Deepskyblue'> Step 4: Create a list of random ingredients to make it fun!

```python
# Define a list of ingredients that will be used to generate a random recipe. These are common household items.
# Imagine a pantry filled with ingredients, and we will be randomly picking from these items to see what we can cook.
ingredients = ["banana", "chocolate", "peanut butter", "bread", "cheese", "apple", "honey", "cinnamon", "tomato", "milk"]
```

# <font color='Deepskyblue'>  Step 5: Pick 3 random ingredients

```python
# We roll the dice to pick three random ingredients from our pantry list.

# Randomly pick three ingredients from the list, making it fun and spontaneous like a cooking challenge.
# This is like a cooking reality show where you are given three random items, and you need to make something tasty out of them.

chosen_ingredients = random.sample(ingredients, 3)

# Print out the randomly chosen ingredients to see what we will use for the recipe.
print(f"Today's random ingredients are: {', '.join(chosen_ingredients)}")
```

#<font color='Deepskyblue'>  Step 6: Use GPT-3 to generate a recipe with the chosen ingredients

```python

# Define a function that takes the chosen ingredients as input and uses GPT-4 to generate a recipe.
# Imagine the AI is a master chef waiting for your instructions to create a recipe.
def promptGPT(prompt, model, system_requirements):
    try:
        # Generate a response from the OpenAI API
        # Here we ask the AI (GPT-4) to generate a recipe based on the ingredients we chose.
        # The 'system' role gives the AI an identity, like telling it "You are a helpful chef," and the 'user' role is like a customer asking for a recipe.
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": system_requirements},
                {"role": "user", "content": prompt}
            ]
        )
        # Return the generated content
        # The AI responds with its creation, which is the recipe in this case.
        return response.choices[0].message.content
    except Exception as e:
        # If something goes wrong, this will return an error message.
        return f"An error occurred: {e}"

# Define the model to use and the system requirements
# We define which version of GPT-4 we want to use, and what kind of assistant it should be.
model = "gpt-4-0125-preview"  # or any other version available to you
system_requirements = "You are a helpful AI designed to create recipes."

# Create a prompt for the AI, asking it to create a recipe based on the chosen ingredients
# We are providing the AI with instructions like "Create a simple and delicious recipe using the following ingredients: ..."
prompt = f"Create a simple and delicious recipe using the following ingredients: {', '.join(chosen_ingredients)}."
```

# <font color='Deepskyblue'> Step 7: Generate and print the recipe

```python
# We call the promptGPT function to generate a recipe based on our selected ingredients.
recipe = promptGPT(prompt, model, system_requirements)
# Print a message to introduce the generated recipe.
print("\nHere is your recipe:")
# Finally, we display the recipe that the AI has created for us.
print(recipe)
```
